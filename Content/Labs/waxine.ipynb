# Waxinekaarsje

Als je een waxinekaarsje aan steekt en er een glas overheen zet, zal het kaarsje uit gaan. 
Als je dat waxinekaarsje laat drijven op water terwijl je er het glas er overheen zet, zal het water in het glas stijgen.
Op het internet zwerven er verschillende verklaringen voor dit fenomeen rond.
1. Het kaarsje verbruikt zuurstof, waardoor er een vacuüm ontstaat en het water omhoog wordt gezogen.
2. Het kaarsje verwarmt de lucht in het glas, waardoor de lucht uitzet. Wanneer de kaars dooft koelt de lucht af waardoor de druk afneemt en er water omhoog wordt gezogen.
3. Bij het verbranden van de kaars ontstaan er waterdamp en koolstofdioxide. De waterdamp condenseert aan de binnenkant van het glas, waardoor er minder gas in het glas is en de druk afneemt. Hierdoor wordt het water omhoog gezogen.

```{exercise}
Onderzoek welke verklaring het meest waarschijnlijk is en leg uit waarom de andere verklaringen minder waarschijnlijk zijn.
```

Er zijn 2 experimenten uitgevoerd: 1. Waar verschillende aantallen waxine lichtjes onder het bekerglas werden gezet. 2. Waar het bekerglas boven het waxinelichtje werd gehouden, vervolgens het waxinelichtjes werd uitgeblazen, en vervolgens het bekerglas erop gezet.

Bij het eerste experiment bleek des te meer waxine lichtjes onder het bekerglas geplaatst werden des te hoger het water niveau steeg.
Uit het tweede experiment bleek des te langer het bekerglas boven het waxinelichtje hangt, des te hoger het water stijgt.

Uit deze resultaten kan worden geconcludeert dat verklaring 2 het meest waarschijnlijk is, omdat bij experiment 2 drukverschil binnen en buiten het bekerglas niet mogelijk is, gezien het niet word afgesloten totdat het waxinelicht uit staat. (alhoewel het mogelijk is dat er enige verbranding nog plaatsvind na het uitblazen zal die hoeveelheid verbranding niet significant genoeg zijn om de resultaten te verklaren) 
Verder sluit experiment 1 verklaring 3 ook uit, gezien ongeacht de hoeveelheid waxinelichtjes, het punt waarop er te weinig zuurstof is om verbranding mogelijk te maken blijft het zelfde. Dus aan het einde van de verbraning bij 4 waxinelichtjes is de hoeveelheid waterdamp/ koolstof dioxide ongeveer evenveel als bij 1 waxine lichtje, waxine lichtje 1 doet er gewoon langer over uit te gaan. De tempratuur zal in beide gevallen echter niet het zelfde zijn, omdat door de snelle opwarming van 4 waxine lichtjes de lcuht binnen het bekerglas minder de mogelijkheid heeft warmte af te staan aan de buitenlucht gedurende de reactie.

Verklaring 1 klopt chemisch gezien niet, gezien de zuurstof die word gebruikt in de verbranding word omgezet in CO2/H2O/CO, dat zou niet resulteren in minder gas, maar meer gas in het bekerglas, gezien de verhoudingen van de reactie:  1.O2 : 1.CO2 / 1.O2 : 2.H2O / 1.O2 : 2.CO

dus de rangorde van waarschijnlijkheid is:
Meest waarschijnlijk: 2,
Middelwaarschijnlijk: 1,
Minstwaarschijnlijk: 3,
