# Bepaling van soortelijke warmte van water


# Introductie

Water heeft een enorme buffercapaciteit: je moet veel energie toevoegen om de temperatuur van water een graad te verwarmen. In dit practicum gaan we de soortelijke warmte van water bepalen door een bekende hoeveelheid water te verwarmen met een bekende hoeveelheid energie, en de temperatuurstijging te meten.

# Theorie

Materialen hebben een bepaalde hoeveelheid energie nodig om een bepaald gewicht een bepaald aantal graden te verwarmen, dat is een unieke constante voor ieder materiaal. Die constante kan worden bepaald door een bekende hoeveelheid energie toe te voegen aan een bekend gewicht van een bepaald materiaal, om vervolgens het verschil in tempratuur te meten. gezien de formule Q = m * c * dt

# Methode en materialen

## Ontwerp
Een waterbad met bekende massa aan water wordt verwarmd met een elektrisch verwarmingselement dat een bekende hoeveelheid energie levert.
De temperatuur van het water wordt gemeten met een temperatuursensor.
Door de temperatuurstijging als functie van de tijd te meten kan de soortelijke warmte van water worden berekend.

## Procedure
```{exercise}
# Procedure voor het bepalen van de soortelijke warmte van water

1. Voorbereiding
   - Vul een maatbeker met een bekende hoeveelheid water (bijvoorbeeld 500 g = 0,5 kg).
   - Meet de massa van het water nauwkeurig met een weegschaal.
   - Plaats de thermometer of temperatuursensor in het water en noteer de begintemperatuur.

2. **Opstelling**
   - Plaats het elektrisch verwarmingselement volledig in het water.
   - Sluit het verwarmingselement aan op de voedingsbron.
   - Zorg dat een roerder (indien beschikbaar) aanstaat om de warmte gelijkmatig te verdelen.

3. Verwarming
   - Zet de voedingsbron aan en stel het vermogen in (bijvoorbeeld 40 W).
   - Start tegelijkertijd de stopwatch.
   - Verwarm het water gedurende een vooraf bepaalde tijd (bijvoorbeeld 5 minuten).
   - Noteer de temperatuur van het water elke 30 seconden.

4. Afronding
   - Zet de voedingsbron uit na de gekozen tijd.
   - Noteer de eindtemperatuur van het water.
   - Bereken de totale energie die is toegevoerd:
     
     Q = P * t
     
     waarbij P het elektrisch vermogen is (in watt) en t de tijd (in seconden).

5. Berekening van soortelijke warmte
   - Gebruik de formule:
     
     c = Q/(m * dT)
     
     waarbij:
       - Q = toegevoerde energie (in joule),
       - m = massa van het water (in kg),
       - Delta T = temperatuurstijging (in °C).
```

# Resultaten


````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit 
T_init = 21.5
m_water = 0.600
vol = 18.1
amp = 1.86
vermogen = vol * amp
time = np.arange(0, 7.5*61, 30)
Q = vermogen * time 
T = [22.0, 22.4, 22.8, 23.1, 23.5, 23.9, 24.1, 24.4, 24.7, 25.0, 25.3, 25.5, 25.9, 26.1, 26.4, 26.8]

plt.figure()
plt.plot(time, Q)
plt.xlabel('Time(s)')
plt.ylabel('Energie(j)')
plt.show()
plt.figure()
plt.plot(time, T)
plt.xlabel('Time(s)')
plt.ylabel('Temprature(C)')
plt.show()

c = Q / (m_water * (np.subtract(T, T_init)))
uc = np.std(c) / np.sqrt(len(c))
uc3 = np.std(c[3:]) / np.sqrt(len(c[3:]))
print (f"berekende soortelijke warmte van water: {np.mean(c):.1f} +/- {uc:.0f} J/kg*k")
print (f"berekende soortelijke warmte van water zonder de eerste 3 metingen: {np.mean(c[3:]):.1f} +/- {uc3:.0f} J/kg*k")
print ("theoretische soortelijke warmte van water: 4186 J/kg*k,")

# agreement analysis
if (abs(np.mean(c) - 4186) < 2 * uc):
    print("de twee waarden komen overeen")
else:
    print("de waarden komen niet overeen")
if (abs(np.mean(c[3:]) - 4186) < 2 * uc3):
    print("de twee waarden zonder de eerste 3 metingen komen overeen")
else:
    print("de waarden zonder de 3 eerste metingen komen niet overeen")
````
